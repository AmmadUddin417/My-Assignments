## 11. Write a program to calculate the sum of even and odd numbers and print them. (numbers from 1 to 50).


```python
#for Loop:

even_sum = 0
odd_sum = 0

for i in range(1, 51):
    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```

```python
#While Loop:

even_sum = 0
odd_sum = 0
i = 1

while i <= 50:
    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

    i = i + 1

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```

## 12. You are organizing a party and need to enter the names of 10 guests. The program shouldcheck whether a guest's name has already been entered. Write a Python program that uses a for loop, list, input(), and if-else to manage a guest list and prevent duplicate names. 


```python
guests = []

for i in range(10):
    name = input("Enter guest name: ")

    if name in guests:
        print("This guest is already added.")
    else:
        guests.append(name)
        print("Guest added.")

print("Guest List:")
print(guests)
```

## 13. You are creating a restaurant menu. The program should allow the user to enter 5 dish names. However, the restaurant does not serve fish, so the program should skip it whenever the user enters "fish". Write a Python program that uses a for loop, if condition, continue, and a list to create a restaurant menu while excluding a specific dish. 


```python
menu = []

while len(menu) < 5:
    dish = input("Enter dish name: ")

    if dish.lower() == "fish":
        print("Fish is not available.")
        continue

    menu.append(dish)

print("Restaurant Menu:")
print(menu)
```

## 14. Aap ek dawat ke liye menu aur uski cost calculate karne wala Python program banayein. Dawat mein 300 mehman hain. User se 5 dishes ke naam lene hain. Lekin kuch dishes menu mein allowed nahi hain. Requirements:
o discard_menu naam ki ek list banayein jismein ye dishes hon:
1. fish
2. qeema
3. nihari
o loop ki madad se user se 5 dishes ke naam input lein.
o Agar user aisi dish enter kare jo discard_menu mein mojood ho:
1. Us dish ko menu mein add na karein.
2. continue ka use karke next dish par chale jayein.
o Agar dish allowed ho to usay menu list mein add karein.
o Har allowed dish ke liye 300 mehmanon ka total price user se input lein.
o Har dish ki price ko menu_price dictionary mein store karein.
o Loop complete hone ke baad tamam dishes ki prices ko add karke total menu price
calculate karein.
o Har dish ki per head cost calculate karein:
 Per Head Cost = Total Dish Price / Number of Guests
o Har dish ki per head cost ko menu_price_per_head dictionary mein store karein.
o Aakhir mein tamam dishes ki per head cost ko add karke complete menu ki per
head cost display karein.

```python
guests = 300

discard_menu = ["fish", "qeema", "nihari"]

menu = []
menu_price = {}
menu_price_per_head = {}

while len(menu) < 5:
    dish = input("Enter dish name: ").lower()

    if dish in discard_menu:
        print("This dish is not allowed.")
        continue

    menu.append(dish)

    price = int(input("Enter total price for 300 guests: "))
    menu_price[dish] = price

total_menu_price = 0

for dish in menu:
    total_menu_price = total_menu_price + menu_price[dish]

    per_head_cost = menu_price[dish] / guests
    menu_price_per_head[dish] = per_head_cost

complete_per_head_cost = 0

for dish in menu_price_per_head:
    complete_per_head_cost = complete_per_head_cost + menu_price_per_head[dish]

print("\nMenu:")
print(menu)

print("\nDish Prices:")
print(menu_price)

print("\nTotal Menu Price:", total_menu_price)

print("\nPer Head Cost of Each Dish:")
print(menu_price_per_head)

print("\nComplete Menu Per Head Cost:", complete_per_head_cost)
```
